```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/placement.csv')
df.head()
```

```python
sns.scatterplot(x='cgpa', y='resume_score', data=df, hue='placed')
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Perceptron

from sklearn.metrics import classification_report, confusion_matrix
```

```python
X_train, X_test, y_train, y_test = train_test_split(df[['cgpa', 'resume_score']], df['placed'], test_size=0.2, random_state=42)

model = Perceptron()
```

```python
model.fit(X_train, y_train)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X_train.values, y_train.values, clf=model, legend=2)
```
